```python
# Importer les fonctions de création de la base.
from utils.settings import datas_folder
from utils.database import load_data, build_events, create_index, save_database

output_dir = datas_folder / "faiss_events"
```

```python
# Charger le CSV et les embeddings déjà calculés dans le notebook 2.
df, embeddings = load_data(datas_folder)

print(f"Événements : {len(df):,}")
print(f"Embeddings : {embeddings.shape[0]:,}")
print(f"Dimension des vecteurs : {embeddings.shape[1]}")
```

```python
# Associer chaque position aux métadonnées de son événement.
events = build_events(df)
print(f"Métadonnées associées : {len(events):,}")
print(events[0]["metadata"])
```

```python
# Créer FAISS avec une copie normalisée des vecteurs existants.
index = create_index(embeddings)
print(f"Vecteurs indexés : {index.ntotal:,}")
print(f"Dimension : {index.d}")
```

```python
# Sauvegarder l'index, les métadonnées et la configuration.
save_database(index, events, output_dir)
print("Base sauvegardée dans :", output_dir)
```
